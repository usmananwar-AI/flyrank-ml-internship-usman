## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
import os

if not os.path.exists('flyrank-ml-internship-starter'):
    !git clone https://github.com/flyrank-bih/flyrank-ml-internship-starter.git

file_path = 'flyrank-ml-internship-starter/data/raw/content_refresh_anonymized.csv'
data = pd.read_csv(file_path)

# Calculate 3 real numbers using the correct column names
total_rows = len(data)
avg_search_volume = data['search_volume'].mean()
decline_rate = (data['impressions_last_30d'] < 0.8 * data['impressions_prev_30d']).mean() * 100

print(f"1. Total content records analyzed: {total_rows:,}")
print(f"2. Average monthly search volume per asset: {avg_search_volume:.2f}")
print(f"3. Percentage of content showing impression decline: {decline_rate:.1f}%")
```
