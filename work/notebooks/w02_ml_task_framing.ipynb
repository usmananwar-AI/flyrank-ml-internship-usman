## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

The Limitation of Fixed Rules: A rigid rule (e.g., "If dwell time < 30 seconds, trigger retention alert") fails because user search intent and context vary wildly. A short session for a navigational query is normal, whereas a short session for an informational query indicates churn.

Why ML Wins: Machine learning dynamically weighs multiple interacting features simultaneously—such as historical search telemetry, query patterns, and interaction depth—to adaptively predict drop-off risk with high precision.
